# Deep Learning — MLP for House-Price Regression

This notebook is a launcher and experiment guide for `dl_mlp_experiment.py`. It evaluates the same MLP pipeline on both assigned datasets.

**Before running:** download the two CSV files from Kaggle and place them under `data/` (or edit the file paths below). The notebook does not fabricate performance results; all numbers and figures are generated by your local run.

Protocol: outer split 80/20 with seed 42; five-fold CV on the training portion; seed 36 for shuffling/model randomness; a validation subset for early stopping; MAE/RMSE/R² on original price units; test set reserved for final evaluation only.

```python
from pathlib import Path
import subprocess
import sys

HOUSING_CSV = Path('data/Housing.csv')
KING_COUNTY_CSV = Path('data/kc_house_data.csv')
OUTPUT_DIR = Path('results_mlp')

assert Path('dl_mlp_experiment.py').exists(), 'Put this notebook next to dl_mlp_experiment.py.'
assert HOUSING_CSV.exists(), f'Missing dataset: {HOUSING_CSV}'
assert KING_COUNTY_CSV.exists(), f'Missing dataset: {KING_COUNTY_CSV}'

command = [
    sys.executable, 'dl_mlp_experiment.py',
    '--housing_csv', str(HOUSING_CSV),
    '--king_county_csv', str(KING_COUNTY_CSV),
    '--output_dir', str(OUTPUT_DIR),
]
print('Running:', ' '.join(command))
subprocess.run(command, check=True)
```

## Where to find the outputs

Open `results_mlp/combined_test_summary.csv` for final test metrics on both datasets. For each dataset folder, inspect `cv_results.csv`, `cv_fold_details.csv`, `best_hyperparameters.json`, `test_predictions.csv`, `final_training_history.csv`, `validation_permutation_importance.csv`, and the `figures/` directory.
